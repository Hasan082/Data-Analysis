```python
import pandas as pd

# Sample DataFrame
data = {
    'Country': ['USA', 'Canada', 'Germany', 'France', 'India'],
    'Population': [331, 38, 83, None, 1393],
    'GDP': [21.4, 1.8, None, 2.7, 2.9],
    'HDI': [0.926, 0.929, 0.947, 0.901, None],
    'Life Expectancy': [78.8, 82.3, 81.0, 83.1, 69.7]
}

df = pd.DataFrame(data)

# Step 1: Identify columns with 3 or fewer missing values
valid_columns = df.columns[df.isna().sum() <= 3]

# Step 2: Remove rows (countries) with missing values in these columns
df_filtered = df.dropna(subset=valid_columns)

print(df_filtered)
```
