# Pie Charts

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('survey-data.csv')
df.head(5)
```

# Visualizing Data Composition with Pie Charts¶
1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With
In the survey data, the DatabaseWantToWorkWith column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.

```python
df_copied = df[['DatabaseWantToWorkWith']].copy()
df_copied_cleaned = df_copied.dropna(subset=['DatabaseWantToWorkWith'])
df_copied_cleaned.loc[:,'DatabaseWantToWorkWith'] = df_copied_cleaned['DatabaseWantToWorkWith'].str.split(';')
df_copied_cleaned = df_copied_cleaned.explode('DatabaseWantToWorkWith')
df_copied_cleaned_count = df_copied_cleaned.value_counts().reset_index()
df_copied_cleaned_count.columns = ['DatabaseWantToWorkWith', 'Count']
df_copied_cleaned_count = df_copied_cleaned_count.sort_values(by='Count', ascending=False)
top_5_databases = df_copied_cleaned_count.sort_values(by='Count', ascending=False).head(5)
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    top_5_databases['Count'],
    labels=top_5_databases['DatabaseWantToWorkWith'],
    autopct='%1.1f%%',
    startangle=90,    
)
plt.axis('equal')
plt.show()
```
